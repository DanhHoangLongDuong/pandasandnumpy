```python
import numpy as np
import pandas as pd
```

set the index with "index_col" 

change index date column to a datetime data type by parse_dates=True \
*dtype('<M8[ns]') is the synonym for datetime64* 

Can select the columns to read in with the "use_cols" argument \
This can save a lot of processing time and memory

```python
columns = ["Date", "Oil Price"]
oil = pd.read_csv("../retailFiles/oil.csv",
                  header=0, 
                  names=columns)
oil.head()
```

```python
columns = ["Date", "Oil Price"]
oil_1 = pd.read_csv("../retailFiles/oil.csv",
                  header=0, 
                  names=columns,
                  index_col="Date")
oil_1.head()
oil_1.index.dtype
```

```python
columns = ["Date", "Oil Price"]
oil_2 = pd.read_csv("../retailFiles/oil.csv",
                  header=0, 
                  names=columns,
                  index_col="Date",
                  parse_dates=True)
oil_2.head()
oil_2.index.dtype
```
