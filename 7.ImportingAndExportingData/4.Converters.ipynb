```python
import pandas as pd
```

Can apply functions to columns of data by using the "converters" argument\
pass a dictionary with the column names as keys and the functions as values

```python
columns = ["Date", "Oil Price"]
oil = pd.read_csv("../retailFiles/oil.csv",
                  header=0,
                  names=columns,
                  index_col="Date",
                  parse_dates=True,
                  usecols=["Date", "Oil Price"],
                  skiprows=[0], # Skip the first row
                  nrows=3, # Use the first 3 rows
#                 skiprows = lambda x: x % 2 == 1   Skip every odd rows
                  )
oil.head()
```

```python
columns = ["Date", "Oil Price"]
string_slice = lambda x: int(x[2:4]) #extract only 2 last numbers of years
oil_1 = pd.read_csv("../retailFiles/oil.csv",
                  header=0,
                  names=columns,
                  index_col="Date",
                  parse_dates=True,
                  converters={"Date": string_slice} # CONVERTERS
                  )
oil_1.head()
```
