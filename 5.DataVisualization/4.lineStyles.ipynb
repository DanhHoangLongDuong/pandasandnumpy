```python
import pandas as pd
```

add chart title and axis labels to increase understanding

```python
transactions = pd.read_csv("../retailFiles/transactions.csv")
transactions.head()
```

```python
transactions_44 = transactions.loc[transactions["store_nbr"] == 44, ["date", "transactions"]]
```

```python
transactions_44.set_index('date').plot()
```

```python
transactions_44.set_index('date').plot(
    title="Store 44 Transaction 2013-2017",
    xlabel="Date",
    ylabel="Daily Transactions"
)
```

```python
transactions_44.set_index('date').plot(
    title="Store 44 Transaction 2013-2017",
    xlabel="Date",
    ylabel="Daily Transactions",
    color="red"
)
```

```python
(transactions
    .query("store_nbr in [44, 47]")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        color=["red", "blue"]
    )
);
```

```python
(transactions
    .query("store_nbr in [44, 47]")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        cmap='tab10', #cmap is built-in set of colors by matplotlib
        alpha=0.2 #make graph transparent, range from 0 (invisible) to 1 (fully visible)
    )
);
```

```python
(transactions
    .query("store_nbr in [44, 47] and date.astype('datetime64[ns]').dt.year == 2014")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        style= ["-", "--"]
    )
);
```
