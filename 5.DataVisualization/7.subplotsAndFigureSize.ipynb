```python
import pandas as pd
```

add chart title and axis labels to increase understanding

```python
transactions = pd.read_csv("../retailFiles/transactions.csv")
transactions.head()
```

```python
(transactions
    .query("store_nbr in [44, 47] and date.astype('datetime64[ns]').dt.year == 2014")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        style= ["-", "--"],
        grid=True
    )
).legend(bbox_to_anchor=(1,1))
```

```python
import seaborn as sns
sns.set_style("darkgrid")
```

```python
#Apply seaborn "darkgrid" style
(transactions
    .query("store_nbr in [44, 47] and date.astype('datetime64[ns]').dt.year == 2014")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        title="Store 44 and 47 Transactions 2014",
        xlabel="Date",
        ylabel="Daily Transactions"
    )
)
```

```python
import matplotlib
matplotlib.style.use("fivethirtyeight")
```

```python
#Apply matplotlib "fivethirtyeight" style
(transactions
    .query("store_nbr in [44, 47] and date.astype('datetime64[ns]').dt.year == 2014")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        title="Store 44 and 47 Transactions 2014",
        xlabel="Date",
        ylabel="Daily Transactions"
    )
)
```

```python
#Apply matplotlib "fivethirtyeight" style
(transactions
    .query("store_nbr in [44, 47] and date.astype('datetime64[ns]').dt.year == 2014")
    .pivot_table(index="date", columns="store_nbr")
    .droplevel(0, axis=1)
    .plot(
        title="Store 44 and 47 Transactions 2014",
        xlabel="Date",
        subplots=True, #Split
        layout=(1, 2), #Side by side
        sharey=True, #Make both graphs share the same y-axis (easier to compare)
        figsize=(10, 10) #increase chart size
    )
)
```
