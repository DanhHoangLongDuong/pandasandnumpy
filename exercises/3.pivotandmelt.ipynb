```python
import numpy as np
import pandas as pd

premier_league = pd.read_excel("../retailFiles/premier_league_games_full.xlsx")
premier_league.head()
```

## Exercise 1 — Your First Pivot Table
*(covers: PivotTables — index / columns / values / aggfunc)*

1. Build a pivot table with **one row per home team**, **one column per season**, and each cell holding the **total home goals** that team scored at home in that season.
2. Now build the same table, but only for three teams: `"Liverpool"`, `"Manchester United"`, and `"Tottenham Hotspur"`. Save it as `three_teams`.

```python
# 1. Total home goals by home team (rows) and season (columns)
premier_league.pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           )

# 2. Same table, limited to three teams -> three_teams
premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        values="HomeGoals",
                                                                                                        aggfunc="sum")
```

## Exercise 2 — Row & Column Totals
*(covers: PivotTables — margins)*

Rebuild the three-team pivot from Exercise 1, but this time add a **row of totals and a column of totals**, so you can read each team's grand total home goals and each season's total across those teams.

```python
# Three-team pivot with row and column totals
premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        values="HomeGoals",
                                                                                                        aggfunc="sum",
                                                                                                        margins=True
                                                                                                        )
```

## Exercise 3 — Two Aggregations at Once
*(covers: PivotTables — list of aggfuncs)*

For the same three teams, build a pivot table (rows = home team, columns = season) where each cell shows **both the sum and the mean** of home goals — in a single table, not two separate ones.

```python
# Sum AND mean of home goals per team per season
premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        values="HomeGoals",
                                                                                                        aggfunc=["sum", "mean"])
```

## Exercise 4 — Aggregating Multiple Value Columns
*(covers: PivotTables — dict aggfunc)*

1. For the three teams, build a pivot table (rows = home team, columns = season) that shows the **sum of `HomeGoals`** *and* the **sum of `AwayGoals`** side by side.
2. Do it again, but this time take the **sum of `HomeGoals`** and the **mean of `AwayGoals`** — each value column with its own aggregation.

```python
# 1. Sum of HomeGoals and sum of AwayGoals
premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        aggfunc={"HomeGoals":"sum", "AwayGoals":"sum"})

# 2. Sum of HomeGoals and mean of AwayGoals
premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        aggfunc={"HomeGoals":"sum", "AwayGoals":"mean"})
```

## Exercise 5 — Turn a Pivot Table into a Heatmap
*(covers: PivotTables — .style.background_gradient)*

1. Build a pivot table of **total home goals** for the three teams (rows = home team, columns = season) and save it as `goals_pivot`.
2. Style `goals_pivot` as a **heatmap** using a red-yellow-green colour map, with the gradient applied **across each row** so you can see each team's best and worst seasons.

```python
# 1. goals_pivot: total home goals, three teams
goals_pivot=premier_league.query("HomeTeam in ['Liverpool', 'Manchester United', 'Tottenham Hotspur']").pivot_table(index="HomeTeam",
                                                                                                        columns="season",
                                                                                                        values="HomeGoals",
                                                                                                        aggfunc="sum")

# 2. Heatmap: red-yellow-green gradient across each row
goals_pivot.style.background_gradient(cmap="RdYlGn", axis=1)
```
