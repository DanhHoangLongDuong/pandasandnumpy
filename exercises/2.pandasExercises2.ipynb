## Exercise 6 — Missing Data, apply() & where()
*(covers: fixingMissingData, practiceFixingMissingData, ApplyingCustomFunctions, where)*

1. Create `borrows_with_nan = pd.Series([12.0, 5.0, np.nan, 0.0, 20.0], index=["fiction", "scifi", "comic", "biography", "romance"])`.
2. Find which values are missing, and how many are missing in total.
3. From `borrows_with_nan`, create two new Series: one with the missing rows removed entirely (`borrows_dropped`), and one with the missing values replaced by 0 (`borrows_filled`).
4. Create `titles2 = pd.Series(["fiction", "scifi", "comic"])`. Using a one-line anonymous function applied to every element, produce the last letter of each title.
5. Write a function `is_popular(genre)` that returns `"Popular"` if the genre is `"fiction"` or `"scifi"`, and `"Available"` otherwise. Apply it to every element of `titles2`.
6. Using `borrows_filled` from step 3, build a new Series where any value above 15 is replaced with half its value (rounded to 2 decimals) — but every other value stays exactly as it was.

```python
borrows_with_nan = pd.Series([12.0, 5.0, np.nan, 0.0, 20.0], index=["fiction", "scifi", "comic", "biography", "romance"])

# 2. Which values are missing, and how many
borrows_with_nan.isna()
borrows_with_nan.isna().sum()
# 3. borrows_dropped (missing rows removed) and borrows_filled (missing values -> 0)
borrows_dropped = borrows_with_nan.dropna()
borrows_filled = borrows_with_nan.fillna(0)
borrows_dropped
borrows_filled
# 4. titles2: last letter of each title, via a one-line anonymous function
titles2 = pd.Series(["fiction", "scifi", "comic"])
titles2.apply(lambda x: x[-1])

# 5. Write is_popular(genre), then apply it to titles2
def is_popular(genre):
    if genre == "fiction" or genre == "scifi":
        return "Popular"
    else:
        return "Available"
titles2.apply(is_popular)


# 6. borrows_filled with anything above 15 halved (rounded to 2 decimals), everything else unchangedaaa
borrows_filled.where(~(borrows_filled > 15), round((borrows_filled / 2), 2))
```

## Exercise 7 — Putting It Together
*(covers: a mixed review of everything above)*

1. Create `ratings = pd.Series([4.5, np.nan, 3.0, np.nan, 5.0, 2.5], index=["fiction", "scifi", "comic", "biography", "romance", "mystery"], name="Ratings")`.
2. Find how many ratings are missing, and which genres they belong to.
3. Create `ratings_filled`, where each missing rating is replaced with the **average of the ratings that do exist** — not a fixed number you type in yourself.
4. Write a function `rating_label(score)` that returns `"Great"` for a score of 4 or above, `"Okay"` for a score from 2.5 up to (but not including) 4, and `"Poor"` for anything below 2.5. Apply it to every value in `ratings_filled` to create a new Series called `rating_labels`.
5. From `ratings_filled`, build a Series where every rating stays as-is, except any rating below 3 gets bumped up to exactly 3 (a "minimum rating" rule).
6. Sort `ratings_filled` from highest to lowest, and show only the top 3 genres.

```python
# 1. Create the ratings Series
ratings = pd.Series([4.5, np.nan, 3.0, np.nan, 5.0, 2.5], index=["fiction", "scifi", "comic", "biography", "romance", "mystery"], name="Ratings")

# 2. Which ratings are missing, and which genres they belong to
ratings.isna()
ratings.isna().sum()
ratings.isna().loc[ratings.isna() == True].index
# 3. ratings_filled: missing values replaced with the average of the existing ratings
ratings_filled = ratings.fillna(np.average(ratings.dropna()))
ratings_filled
# 4. Write rating_label(score), then apply it to ratings_filled -> rating_labels
def rating_label(score):
    if score >= 4:
        return "Great"
    elif score >= 2.5 and score < 4:
        return "Okay"
    else:
        return "Poor"
rating_labels = ratings_filled.apply(rating_label)
rating_labels
# 5. ratings_filled with anything below 3 bumped up to exactly 3
ratings_filled.where(~(ratings_filled < 3), 3)

# 6. Top 3 genres by rating, highest to lowest
ratings_filled.sort_values(ascending=True).iloc[-3:].index
```
