```python
import numpy as np
import pandas as pd
```

sort a DataFrame by its indices using the .sort_index() method (sorts rows axis=0 by default, can specify axis=1 to sort the columns)

sort a DataFrame by its values using the .sort_values() nethod
can sort by a single column or by multiple columns

```python
oil = pd.read_csv("../retailFiles/oil.csv")
oil["benchmark"] = 100
oil
```

```python
oil.sort_index(ascending=False)
```

```python
oil.sort_index(ascending=False, axis=1)
```

```python
oil["month"] = oil["date"].astype("datetime64[ns]").dt.month
oil
```

```python
oil.sort_values("dcoilwtico")
```

```python
oil.sort_values("dcoilwtico", ascending=False)
```

```python
oil.sort_values(["month", "dcoilwtico"], ascending=[True, False])
```
