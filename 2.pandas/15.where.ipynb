```python
import numpy as np
import pandas as pd
```

```python
clean_wholesale = pd.Series([3.99, 5.99, 22.99, 7.99, 33.99])
clean_wholesale
```

```python
#.where(logical test, -> Evaluates to True or False
# value if False, -> Value to return when the expression is False
# implace=False) -> Whether to perform the operation in place (default is False)

# Use a tilde `~` to invert the Boolean values and turn this into a "value if True"
clean_wholesale.where(~(clean_wholesale > 20), round(clean_wholesale * 0.9, 2))
```

```python
(clean_wholesale
    .where(~(clean_wholesale > 20), round(clean_wholesale * 0.9, 2))
    .where(clean_wholesale > 10, 0)
)
```

```python
np.where(clean_wholesale > 20, "Discounted", "Normal Price")
```
