Can append, or concatenate, DataFrames with the concat() function\
The columns for the DataFrames should be identical\
pd.concat([df_1, df_2]) will stack the rows from df_2 at the bottom of df_1

```python
import pandas as pd
```

```python
retail = pd.read_csv("../retailFiles/retail_2016_2017.csv")
retail.head()
```

```python
automotive = retail.query("family == 'AUTOMOTIVE'").head()
automotive
```

```python
beauty = retail.query("family == 'BEAUTY'").head()
beauty
```

```python
pd.concat([automotive, beauty])
```

```python
pd.concat([automotive, beauty],
          ignore_index=True)
```
