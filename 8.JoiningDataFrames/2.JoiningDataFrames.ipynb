### Inner Joins

```python
transactions = pd.read_csv("../retailFiles/transactions.csv", parse_dates=["date"])
oil = pd.read_csv("../retailFiles/oil.csv", parse_dates=["date"])
```

```python
transactions.info()
```

```python
oil.info()
```

```python
transactions.merge(oil,
                   how="inner",
                   left_on=["date"],
                   right_on=["date"])
```

### Left Joins

```python
left_joins_table = transactions.merge(oil,
                                      how="left",
                                      left_on=["date"],
                                      right_on=["date"])
left_joins_table
```

```python
left_joins_table.loc[left_joins_table["date"] == '2013-01-06'].head() # oil prices are all NaN because of oil table
```

### Outer Joins

```python
transactions.merge(oil,
                   how="outer",
                   left_on=["date"],
                   right_on=["date"]).tail()
# transactions table only go up to 2017-08-15, but oil table go up to 2017-08-31
# That is why store_nbr and transactions are all NaN at the end
```
