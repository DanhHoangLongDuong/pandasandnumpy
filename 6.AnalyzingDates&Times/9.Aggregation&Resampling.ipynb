```python
import numpy as np
import pandas as pd
```

Resampling is a special form of time series aggregation that modifies date indices and fills any gaps to creast continuous date values
- Frequencies are used to define the granularity (you can use years, quarters, and months)

```python
retail = pd.read_csv("../retailFiles/retail_2016_2017.csv", parse_dates=["date"])
retail.head()
```

```python
retail.groupby([retail["date"].dt.year, retail["date"].dt.month]).sum(numeric_only=True).round(2)
```

```python
retail.set_index("date").resample("ME").sum(numeric_only=True).round(2)
```

```python
retail.set_index("date").resample("YE").sum(numeric_only=True).round(2)
```

```python
retail.set_index("date").resample("QE").sum(numeric_only=True).round(2)
```
