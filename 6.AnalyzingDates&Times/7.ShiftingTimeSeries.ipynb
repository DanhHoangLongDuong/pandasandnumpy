```python
import pandas as pd
```

.shift() shift a Series by a specifies number of rows

```python
retail = pd.read_csv("../retailFiles/retail_2016_2017.csv", parse_dates=["date"])
retail.head()
```

```python
retail_agg = retail.groupby(["date"]).agg({"sales":"sum"}).round().iloc[1:10]
retail_agg
```

```python
retail_agg.assign(yesterday_sales = retail_agg["sales"].shift())
```

```python
retail_agg.assign(ratio_of_yesterday_sales_and_today_sales = retail_agg["sales"] / retail_agg["sales"].shift())
```
