```python
import pandas as pd
```

.diff() method calculates the different between the values in a Series and those same values shifted a specified number of periods

```python
retail = pd.read_csv("../retailFiles/retail_2016_2017.csv", parse_dates=["date"])
retail.head()
```

```python
retail_agg = retail.groupby(["date"]).agg({"sales":"sum"}).round().iloc[1:10]
retail_agg
```

```python
# dates.diff() = dates - dates.shift(1)
retail_agg.assign(daily_change = retail_agg["sales"].diff())
```

```python
retail_agg.assign(daily_change = (retail_agg["sales"].diff()) / retail_agg["sales"].shift())
```
