```python
import pandas as pd
```

Time delta represent the amount or time, or difference, between two datetimes
- A time delta is returned when subtracting two datetime values

Time delta arithmatic to offset dates by a specified period of time
eg. shipping_dates + pd.to_timedelta(1, unit="D") \
*(2021-04-07 -> 2021-04-08)*

```python
retail = pd.read_csv("../retailFiles/retail_2016_2017.csv", parse_dates=["date"])
retail.head()
```

```python
retail.info(memory_usage="deep")
```

```python
retail = retail.assign(birthday=pd.to_datetime("1950-01-01"))
retail.head()
```

```python
retail = retail.assign(age=retail["date"] - retail["birthday"])
```

```python
retail.head()
```

```python
retail = retail.assign(age=(retail["date"] - retail["birthday"]).dt.days / 365.25)
```

```python
retail.head()
```

```python
retail = retail.assign(retirement_date = retail["date"] +(pd.to_timedelta(365.25, unit="D")) * 20)
retail.head()
```
