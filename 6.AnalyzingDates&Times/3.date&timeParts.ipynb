### Datetime Format Codes

| Category | Directive | Meaning | Example |
|---|---|---|---|
| **Year** | `%Y` | 4-digit year | `2026` |
| | `%y` | 2-digit year (zero-padded) | `26` |
| **Month** | `%m` | Month number (zero-padded) | `09` |
| | `%B` | Full month name | `September` |
| | `%b` | Abbreviated month name | `Sep` |
| **Day** | `%d` | Day of the month (zero-padded) | `23` |
| | `%j` | Day of the year (zero-padded) | `266` |
| **Time** | `%H` | Hour (24-hour clock) | `14` |
| | `%I` | Hour (12-hour clock) | `02` |
| | `%M` | Minute (zero-padded) | `05` |
| | `%S` | Second (zero-padded) | `09` |
| | `%p` | AM or PM | `PM` |
| **Sub-seconds** | `%f` | Microseconds (6 digits) | `123456` |
| | `%N` | Nanoseconds (9 digits, **pandas specific**) | `123456789` |
| **Compound** | `%F` | Shortcut for `%Y-%m-%d` | `2026-09-23` |
| | `%X` | Locale's appropriate time representation | `14:05:09` |
| | `%D` | Shortcut for `%m/%d/%y` | `09/23/26` |
