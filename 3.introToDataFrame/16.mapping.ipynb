```python
import numpy as np
import pandas as pd
```

.map() method maps values to a column or an entire DataFrame
- You can pass a dictionary with existing values as the keys, and new values as the values

```python
retail = pd.read_csv("../retailFiles/retail.csv")
retail.head()
```

```python
retail.loc[:, "family"].value_counts()
```

```python
product_category_dict = {
    'PRODUCE': 'Grocery',
    'POULTRY': 'Grocery',
    'GROCERY I': 'Grocery',
    'GROCERY II': 'Grocery',
    'EGGS': 'Grocery'
}
```

```python
retail.loc[:, "family"].map(product_category_dict)
```

```python
retail.loc[:, "family"].map(product_category_dict).value_counts(dropna=False)
```
