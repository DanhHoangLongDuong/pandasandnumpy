```python
import numpy as np
import pandas as pd
```

```python
retail = pd.read_csv("../retailFiles/retail.csv")
```

.query() method lets you use SQL-like syntax to filter DataFrames
- you can specify any number of filtering conditions by using the "and" & "or" keywords
- you can reference variables by using the "@" symbol

```python
retail.query("family in ['CLEANING', 'DAIRY'] and sales > 0")
```

```python
avg_sales = retail.loc[:, "sales"].mean()
avg_sales
```

```python
retail.query("family in ['CLEANING', 'DAIRY'] and sales > @avg_sales")
```
