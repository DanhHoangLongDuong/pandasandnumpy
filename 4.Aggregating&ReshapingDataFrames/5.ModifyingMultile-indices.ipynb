Several ways to modify multi-index DataFrames:
- Reset the index: Move the index levels back to DataFrame columns (sales_sums.reset_index())
- Swap the index level: Changes the hierarchy for the index levels (sales_sums.swaplevel())
- Drop an index level: Drops an index level from the DataFrame entirely (sales)sums.droplevel('family') (!!! May lose important information)

```python
premier_league = pd.read_excel("../retailFiles/premier_league_games_full.xlsx")
```

```python
agg_prem_league = premier_league.groupby(["season", "HomeTeam"]).agg({"HomeGoals": ['sum', 'mean']})
agg_prem_league.head()
```

```python
agg_prem_league.droplevel(0, axis=1)
```

```python
agg_prem_league = agg_prem_league.droplevel(0, axis=1)
agg_prem_league
```

```python
agg_prem_league.swaplevel()
```

```python
agg_prem_league.swaplevel().loc["Arsenal"]
```

```python
agg_prem_league.reset_index()
```
