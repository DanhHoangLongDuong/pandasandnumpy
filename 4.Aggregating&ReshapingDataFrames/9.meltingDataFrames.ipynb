```python
import numpy as np
import pandas as pd
```

```python
premier_league = pd.read_excel("../retailFiles/premier_league_games_full.xlsx")
premier_league.head()
```

.melt() will unpivot a DataFrame, or convert columns into rows

Use the "id_vars" argument to specify the column to unpivot the DataFrame, "value_vars" to select needed value, "var_name" to name value vars column, "value_name" to name value column

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           )
```

```python
pm = premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           )
```

```python
pm.reset_index().melt()
```

```python
pm.reset_index().melt(id_vars="HomeTeam")
```

```python
pm.reset_index().melt(id_vars="HomeTeam",
                      value_vars=["2008/2009", "2009/2010", "2010/2011"],
                      value_name="avg_goals"
                      )
```
