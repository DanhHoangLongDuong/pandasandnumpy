```python
import numpy as np
import pandas as pd
```

.pivot_table() method let's you create Excel-style pivot tables

- index: returns a row index with distinct values from the specified column
- columns: returns a column index with distinct values from the specified column
- values: the column, or columns, to perform the aggregations on
- aggfunc: defines the aggregation function, or functions, to perform on the "values"
- margins: returns row and column totals when True (False by default)

```python
premier_league = pd.read_excel("../retailFiles/premier_league_games_full.xlsx")
premier_league.head()
```

```python
premier_league.pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           )
```

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           )
```

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum",
                           margins=True
                           )
```

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc=["sum", "mean"]
                           )
```

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           aggfunc={"HomeGoals":["sum", "mean"], "AwayGoals": ["sum"]}
                           )
```

HEATMAPS: chain .style.background_gradient() to the DataFrame and add a "cmap" argument (axis=None applies the gradient across to the whole table, axis=1 applies the gradient across columns, axis=0 applies the gradient across rows)

```python
premier_league.query("HomeTeam in ['Arsenal', 'Chelsea', 'Everton']").pivot_table(index="HomeTeam",
                           columns="season",
                           values="HomeGoals",
                           aggfunc="sum"
                           ).style.background_gradient(cmap="RdYlGn", axis=None)
```
