```python
import numpy as np
import pandas as pd
```

```python
premier_league = pd.read_excel("../retailFiles/premier_league_games_full.xlsx")
premier_league.head()
```

```python
premier_league = premier_league.drop(columns=["league_name", "AwayTeam"])
```

```python
(
    premier_league
    .groupby(["season", "HomeTeam"], as_index=False)
    .agg(["sum", "mean"])
)
```

```python
(
    premier_league
    .groupby(["season", "HomeTeam"], as_index=False)
    .agg({"HomeGoals": ["sum", "mean"],
          "AwayGoals": ["sum", "mean"]
        })
)
```

```python
(
    premier_league
    .groupby(["season", "HomeTeam"], as_index=False)
    .agg(home_goal_sum = ("HomeGoals", "sum"),
         away_goal_sum = ("AwayGoals", "sum")
        )
)
```
